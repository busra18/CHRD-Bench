# 01 - Veri Seti Kontrolü (Dataset Check)

Bu notebook, CHRD (Cultural Heritage Representation Dataset) veri setinin ilk doğrulama işlemlerini gerçekleştirir.

## Amaç

- Proje klasör yapısını kontrol etmek
- Metadata dosyasını okumak
- Görsellerin varlığını doğrulamak
- Metadata alanlarını incelemek
- Eksik veri kontrolü yapmak
- Veri seti hakkında temel istatistikleri oluşturmak

---

## 1. Gerekli Kütüphanelerin Yüklenmesi

Bu bölümde notebook boyunca kullanılacak Python kütüphaneleri içe aktarılır.

```python
from pathlib import Path
import pandas as pd
```

## 2. Proje Dizinlerinin Tanımlanması

Projede kullanılacak klasör ve dosya yolları tanımlanır.

```python
PROJECT_ROOT = Path("..")

DATA_DIR = PROJECT_ROOT / "data"

REAL_DIR = DATA_DIR / "CHRD_Real"

RAW_DIR = REAL_DIR / "raw"

METADATA = REAL_DIR / "metadata.csv"
```

## 3. Proje Klasör Yapısının Kontrolü

Gerekli klasör ve dosyaların mevcut olup olmadığı kontrol edilir.

```python
paths = {
    "Proje Klasörü": PROJECT_ROOT,
    "Veri Klasörü": DATA_DIR,
    "Gerçek Görseller": REAL_DIR,
    "Ham Görseller": RAW_DIR,
    "Metadata": METADATA
}

for name, path in paths.items():
    print(f"{name:<20} {'✅ Bulundu' if path.exists() else '❌ Bulunamadı'}")
```

## 4. Metadata Dosyasının Okunması

Metadata dosyası pandas DataFrame yapısına aktarılır.

```python
metadata = pd.read_csv(METADATA, sep=";")

metadata.head()
```

## 5. Veri Seti Boyutunun İncelenmesi

Toplam satır ve sütun sayısı görüntülenir.

```python
metadata.shape
```

## 6. Metadata Alanlarının İncelenmesi

Metadata dosyasında bulunan sütun isimleri listelenir.

```python
metadata.columns
```

## 7. Eksik Veri Kontrolü

Metadata içerisinde eksik bilgi bulunan alanlar kontrol edilir.

```python
metadata.isnull().sum()
```

## 8. Görsel Sayısının Kontrolü

Veri klasöründeki toplam görsel sayısı hesaplanır.

```python
images = list(RAW_DIR.glob("*"))

print(f"Toplam Görsel Sayısı : {len(images)}")
```

## 9. Metadata ve Görsel Sayısının Karşılaştırılması

Metadata kayıt sayısı ile gerçek görsel sayısının eşleşip eşleşmediği kontrol edilir.

```python
print(f"Metadata Kayıt Sayısı : {len(metadata)}")
print(f"Görsel Sayısı         : {len(images)}")

if len(metadata) == len(images):
    print("✅ Metadata ile görseller eşleşiyor.")
else:
    print("❌ Metadata ile görseller arasında fark var.")
```
