# Promptların CSV Olarak Kaydedilmesi

Bu aşamada promptlar yeni bir CSV dosyası olarak kaydedilmektedir.

Bu dosya daha sonra farklı üretken yapay zekâ modelleri tarafından okunabilecektir.

```python
from pathlib import Path

# prompts klasörünü oluştur (yoksa)
Path("../prompts").mkdir(parents=True, exist_ok=True)

df[
    [
        "image_id",
        "theme",
        "prompt_tr",
        "prompt_en"
    ]
].to_csv(
    "../prompts/prompts.csv",
    index=False,
    encoding="utf-8-sig"
)

print("prompts.csv başarıyla oluşturuldu.")
```

# İngilizce Prompt Dosyasının Oluşturulması

İngilizce promptlar satır satır tek bir metin dosyasına kaydedilmektedir.

Bu dosya İngilizce istemlerle görsel üretimi gerçekleştirmek amacıyla hazırlanmıştır.

```python
with open(
    "../prompts/prompt_en.txt",
    "w",
    encoding="utf-8"
) as f:

    for prompt in df["prompt_en"]:
        f.write(prompt + "\n")

print("prompt_en.txt oluşturuldu.")
```

# Oluşturulan Dosyaların Kontrolü

Notebook sonunda oluşturulan dosyaların ilk birkaç satırı okunarak
kayıt işleminin başarılı olduğu doğrulanmaktadır.

```python
import os

os.makedirs("../prompts", exist_ok=True)
```

```python
with open("../prompts/prompt_tr.txt", "w", encoding="utf-8") as f:
    for prompt in df["prompt_tr"]:
        f.write(str(prompt) + "\n")

print("Dosya oluşturuldu.")
```
